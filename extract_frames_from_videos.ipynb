```python
import cv2
import os
from tqdm import tqdm
```

Specify your absolut video path e.g. C:\Users\AlvinWenJH\Desktop\data\alvin\na_GH011012.MP4_compressed.mp4

```python
input_video_path = r"D:\Tugas Akhir\Subject TA\S34\S34 raw video\S34FRCompressed.mp4"
output_dir = r"D:\Tugas Akhir\Subject TA\S34\S34_FR_Frames"
```

```python
if not os.path.exists(output_dir):
    os.mkdir(output_dir)
    print("Output path not found... creating the specified folder")

cap = cv2.VideoCapture(input_video_path)
j=0
totalFrame = round(cap.get(cv2.CAP_PROP_FRAME_COUNT))
pbar = tqdm(total=totalFrame)
while (cap.isOpened()):
    ret, frame = cap.read()
    if not ret:
        break
    cv2.imwrite(os.path.join(output_dir,f"Frame_{j}.png"),frame)
    j+=1
    # cv2.imshow('Frame', frame)
    if cv2.waitKey(0) & 0xFF == 27:
        break
    pbar.update(1)
cv2.destroyAllWindows()
cap.release()
pbar.close()
```
