```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the csv input
data = pd.read_csv('D:\Tugas Akhir\Subject TA\S43\Yolov7\S43_YOLOV7_KNEE_ANGLE_FILTERED - Copy.csv')

# Calculate the number of frames for the first 12 seconds
frames_for_12_seconds = 12 * 120

# Extract the data only for the first 12 seconds
data_12_seconds = data.iloc[:frames_for_12_seconds]

# Create an index for the x-axis based on the number of frames
x_index = range(frames_for_12_seconds)

# Plot for Knee Angle Raw
plt.figure(figsize=(12, 6))
plt.subplot(2, 1, 1)
plt.plot(x_index, data_12_seconds['knee_angle_raw'], color='blue')
plt.title('Knee Angle Raw over the First 12 Seconds or up to frame 1440 (Filtered) for OpenPose')
plt.ylabel('Knee Angle (degrees)')
plt.xlabel('Frame number')
plt.grid(True)

# Adjusting layout
plt.tight_layout()
plt.show()
```
