```python
import pandas as pd
import numpy as np
from math import sqrt
from scipy.stats import pearsonr
```

```python
# Read the first CSV file (considered as the ground truth)
df_ground_truth = pd.read_csv('D:\Tugas Akhir\Subject TA\S49\Marker\S49_MARKER_KNEE_ANGLE_FILTERED.csv')

# Read the second CSV file
df_comparison = pd.read_csv('D:\Tugas Akhir\Subject TA\S49\OpenPose csv\S49_OPENPOSE_KNEE_ANGLE_FILTERED.csv')
```

```python
# Only consider the first 1440 rows (first 12 seconds at 120 fps)
df_ground_truth = df_ground_truth.iloc[:1440]
df_comparison = df_comparison.iloc[:1440]

# Calculate Pearson correlation for knee_angle_raw
corr_knee_angle_raw, _ = pearsonr(df_ground_truth['knee_angle_raw'], df_comparison['knee_angle_raw'])


# Calculate the squared differences for knee_angle_raw
squared_diffs_raw = (df_ground_truth['knee_angle_raw'] - df_comparison['knee_angle_raw']) ** 2

# Calculate the RMSE for knee_angle_raw
rmse_knee_angle_raw = sqrt(np.mean(squared_diffs_raw))

# Print the Pearson correlation coefficients and RMSE values
print(f"Pearson correlation: {corr_knee_angle_raw}")
print(f"RMSE for knee_angle: {rmse_knee_angle_raw}")
```
