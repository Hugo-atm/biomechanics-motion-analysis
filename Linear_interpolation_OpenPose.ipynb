```python
import numpy as np
import pandas as pd

def linear_interpolate(data):
    """Interpolate zeros in a dataset using linear interpolation."""
    non_zero_indices = np.where(data != 0)[0]
    if len(non_zero_indices) == 0:
        return data
    
    interpolated_data = np.interp(np.arange(len(data)), 
                                  non_zero_indices, 
                                  data[non_zero_indices])
    return interpolated_data

def interpolate_csv_file_linear(input_path, output_path):
    """Load data from a CSV file, interpolate it using linear interpolation, and save to another CSV file."""
    # Load the data from the CSV file
    motion_data = pd.read_csv(input_path)
    
    # Apply the interpolation to each column
    for column in motion_data.columns:
        motion_data[column] = linear_interpolate(motion_data[column].values)
    
    # Save the interpolated data to a new CSV file
    motion_data.to_csv(output_path, index=False)
    print(f"Interpolated data saved to {output_path}")

# Paths for your files
input_folder_linear = 'D:\Tugas Akhir\Subject TA\S49\OpenPose csv\S49_OPENPOSE.csv'
output_folder_linear = 'D:\Tugas Akhir\Subject TA\S49\OpenPose csv\S49_OPENPOSE_INTERPOLATED.csv'

interpolate_csv_file_linear(input_folder_linear, output_folder_linear)
```
