```python
import pandas as pd
import scipy.stats as stats
```

```python
# Load the csv input
data = pd.read_excel(r'C:\Users\hp\Downloads\TA\Final result.xlsx', skiprows=5)
```

```python
data
```

```python
pearson_YOLOv7 = data.loc[0:14, 'Yolov7_Pearson']
pearson_OpenPose = data.loc[0:14, 'OpenPose_Pearson']
RMSE_YOLOv7 = data.loc[0:14, 'Yolov7_RMSE']
RMSE_OpenPose = data.loc[0:14, 'OpenPose_RMSE']
```

```python
pearson_YOLOv7 = pd.to_numeric(pearson_YOLOv7, errors='coerce')
pearson_OpenPose = pd.to_numeric(pearson_OpenPose, errors='coerce')
RMSE_YOLOv7 = pd.to_numeric(RMSE_YOLOv7, errors='coerce')
RMSE_OpenPose = pd.to_numeric(RMSE_OpenPose, errors='coerce')
```

```python
RMSE_OpenPose
```

```python
print(pearson_YOLOv7.dtype)
print(pearson_OpenPose.dtype)
print(RMSE_YOLOv7.dtype)
print(RMSE_OpenPose.dtype)
```

```python
# Performing an independent two-sample t-test
t_statistic, p_value = stats.ttest_ind(pearson_YOLOv7, pearson_OpenPose,alternative='two-sided')

print("t-statistic:", t_statistic)
print("p-value:", p_value)
```

```python
# Performing an independent two-sample t-test
t_statistic, p_value = stats.ttest_ind(RMSE_YOLOv7, RMSE_OpenPose)

print("t-statistic:", t_statistic)
print("p-value:", p_value)
```
