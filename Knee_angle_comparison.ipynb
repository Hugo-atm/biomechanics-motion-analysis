```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
file_paths = ['D:\Tugas Akhir\Subject TA\S43\Marker\S43_MARKER_KNEE_ANGLE_FILTERED.csv', 'D:\Tugas Akhir\Subject TA\S43\Yolov7\S43_YOLOV7_KNEE_ANGLE_FILTERED - Copy.csv', 'D:\Tugas Akhir\Subject TA\S43\OPENPOSE CSV\S43_OPENPOSE_KNEE_ANGLE_FILTERED - Copy.csv']
```

```python
file_paths
```

```python
fps = 120  
duration_seconds = 12  
final_frame = fps * duration_seconds
x_axis = range(final_frame)
```

```python
# Create a figure for plottin, 12))
plt.figure(figsize=(20,6))

# Define labels for legends based on your file names or descriptions
labels = ['Marker method', 'Yolov7', 'OpenPose']  # Replace with appropriate labels if needed

# Loop through each file and plot data
for idx, file_path in enumerate(file_paths):
    df = pd.read_csv(file_path)
    subset_df = df.iloc[:final_frame]

    # Knee angle raw
    plt.subplot(6, 1, 1)
    plt.plot(x_axis, subset_df['knee_angle_raw'], label=labels[idx])
    plt.title("Knee angle (Degrees) vs. Frame", fontsize=25)
    plt.ylabel("Knee angle (Degrees)",fontsize=23)
    plt.xlabel('Frame number',fontsize=23)
        # Here's the additional part for increasing the size of the tick labels
    plt.tick_params(axis='x', labelsize=20)  # Increase x-axis tick label size
    plt.tick_params(axis='y', labelsize=20)  # Increase y-axis tick label size
    plt.grid()
    plt.legend(fontsize=20)
    
plt.tight_layout()
plt.show()
```
