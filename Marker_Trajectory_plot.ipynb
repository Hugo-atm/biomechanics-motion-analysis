```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('D:\Tugas Akhir\Subject TA\S32\OpenPose csv\S32_FR_OPENPOSE_INTERPOLATED_FILTERED.csv') # load the csv file
```

```python
# Constants
fps = 120  # frames per second
duration_seconds = 12  # duration in seconds
final_frame = fps * duration_seconds

# Extract the data for the first 12 seconds
subset_df = df.iloc[:final_frame]

# Define the x indexing
x_index = range(final_frame)

# Plot each marker's x and y coordinates in separate plots against time
plt.figure(figsize=(14, 12))

# Marker 1
plt.subplot(6, 1, 1)
plt.plot(x_index, subset_df["Marker1x"], color='r')
plt.title("Keypoint 1 X-Coordinate (hip) vs. Time", fontsize= 14)
plt.ylabel("X-Coordinate (Pixel)", fontsize=12)

plt.subplot(6, 1, 2)
plt.plot(x_index, subset_df["Marker1y"], color='b')
plt.title("Keypoint 1 Y-Coordinate (hip) vs. Time", fontsize=14)
plt.ylabel("Y-Coordinate (Pixel)", fontsize=12)

# Marker 2
plt.subplot(6, 1, 3)
plt.plot(x_index, subset_df["Marker2x"], color='r')
plt.title("Keypoint 2 X-Coordinate (knee) vs. Time",fontsize=14)
plt.ylabel("X-Coordinate (Pixel)", fontsize=12)

plt.subplot(6, 1, 4)
plt.plot(x_index, subset_df["Marker2y"], color='b')
plt.title("Keypoint 2 Y-Coordinate (knee) vs. Time",fontsize=14)
plt.ylabel("Y-Coordinate (Pixel)", fontsize=12)

# Marker 3
plt.subplot(6, 1, 5)
plt.plot(x_index, subset_df["Marker3x"], color='r')
plt.title("Keypoint 3 X-Coordinate (ankle) vs. Time", fontsize=14)
plt.ylabel("X-Coordinate (Pixel)",fontsize=12)

plt.subplot(6, 1, 6)
plt.plot(x_index, subset_df["Marker3y"], color='b')
plt.title("Keypoint 3 Y-Coordinate (ankle) vs. Time", fontsize=14)
plt.ylabel("Y-Coordinate (Pixel)",fontsize=12)
plt.xlabel("Frame number",fontsize=12)

plt.tight_layout()
plt.show()
```
