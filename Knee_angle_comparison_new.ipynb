```python
import pandas as pd
import matplotlib.pyplot as plt

# Define the file paths (You'll need to update these with the correct paths)
hsv_file_path = 'D:\Tugas Akhir\Subject TA\S32\Marker\S32_FR_MARKER_KNEE_ANGLE_FILTERED.csv'
yolov7_file_path = 'D:\Tugas Akhir\Subject TA\S32\YOLOV7\S32_FR_YOLOV7_KNEE_ANGLE_FILTERED.csv'
openpose_file_path = 'D:\Tugas Akhir\Subject TA\S32\OpenPose csv\S32_FR_OPENPOSE_KNEE_ANGLE_FILTERED.csv'

# Load the CSV files
hsv_data = pd.read_csv(hsv_file_path)
yolov7_data = pd.read_csv(yolov7_file_path)
openpose_data = pd.read_csv(openpose_file_path)

# Calculate the number of frames for the first 12 seconds
frames_for_12_seconds = 12 * 120  # Assuming 120 fps for all

# Extract the data only for the first 12 seconds
hsv_data_12_seconds = hsv_data.iloc[:frames_for_12_seconds]
yolov7_data_12_seconds = yolov7_data.iloc[:frames_for_12_seconds]
openpose_data_12_seconds = openpose_data.iloc[:frames_for_12_seconds]

# Create an index for the x-axis based on the number of frames
x_index = range(frames_for_12_seconds)

# Plot for Knee Angle Comparison
plt.figure(figsize=(20, 6))

# Plot for HSV Motion Capture
plt.plot(x_index, hsv_data_12_seconds['knee_angle_raw'], color='red', label='HSV Motion Capture')

# Plot for YOLOv7
plt.plot(x_index, yolov7_data_12_seconds['knee_angle_raw'], color='green', label='YOLOv7')

# Plot for OpenPose
plt.plot(x_index, openpose_data_12_seconds['knee_angle_raw'], color='blue', label='OpenPose')

plt.title('Knee Angle (Degrees) vs frame',fontsize=20)
plt.ylabel('Knee Angle (Degrees)',fontsize=18)
plt.xlabel('Frame',fontsize=18)
# Here's the additional part for increasing the size of the tick labels
plt.tick_params(axis='x', labelsize=20)  # Increase x-axis tick label size
plt.tick_params(axis='y', labelsize=20)  # Increase y-axis tick label size
plt.legend(fontsize='x-large')
plt.grid(True)

# Adjusting layout
plt.tight_layout()
plt.show()
```
