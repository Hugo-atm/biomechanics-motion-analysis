```python
import pandas as pd
from scipy import signal
import os
```

```python
def filter_motion_data():
    # input csv file
    input_path = r'D:\Tugas Akhir\Subject TA\S49\YOLOV7\S49FRMARKERLESSCOMPRESSEDnotTrimmed.csv'
    data_in = pd.read_csv(input_path)
    
    # cut off frequency
    fc = float(input("Please enter the desired cutoff frequency (e.g., 10): "))
    
    # Butterworth filter parameters
    fs = 120  # Sampling frequency
    n = 4     # Filter order
    b, a = signal.butter(n, 2*fc/fs)

    # Filtering each column of the data
    for column in data_in.columns:
        data_in[column] = signal.filtfilt(b, a, data_in[column])
    
    # 3. Output path
    output_path = r'D:\Tugas Akhir\Subject TA\S49\YOLOV7\S49_YOLOV7_FILTERED.csv'
    
    # 4. Save the filtered data to the specified output path in CSV format
    data_in.to_csv(output_path, index=False)
    print(f"Filtered data saved to: {output_path}")

# Call the function
filter_motion_data()
```
