```python
import numpy as np
import pandas as pd
from scipy.interpolate import CubicSpline

def cubic_spline_interpolate(data):
    """Interpolate zeros in a dataset using cubic spline."""
    non_zero_indices = np.where(data != 0)[0]
    if len(non_zero_indices) == 0:
        return data
    cs = CubicSpline(non_zero_indices, data[non_zero_indices])
    interpolated_data = data.copy()
    zero_indices = np.where(data == 0)[0]
    valid_zero_indices = zero_indices[(zero_indices > non_zero_indices[0]) & (zero_indices < non_zero_indices[-1])]
    interpolated_data[valid_zero_indices] = cs(valid_zero_indices)
    return interpolated_data

def interpolate_csv_file(input_path, output_path):
    """Load data from a CSV file, interpolate it, and save to another CSV file."""
    # Load the data from the CSV file
    motion_data = pd.read_csv(input_path)
    
    # Apply the interpolation to each column
    for column in motion_data.columns:
        motion_data[column] = cubic_spline_interpolate(motion_data[column].values)
    
    # Save the interpolated data to a new CSV file
    motion_data.to_csv(output_path, index=False)
    print(f"Interpolated data saved to {output_path}")

if __name__ == "__main__":
    # Input and output folders
    input_folder = r'D:\Tugas Akhir\Subject TA\S34\S34 OPENPOSE CSV\S34_OPENPOSE.csv'
    output_folder = r'D:\Tugas Akhir\Subject TA\S34\S34 OPENPOSE CSV\S34_OPENPOSE_INTERPOLATED.csv'
    
    interpolate_csv_file(input_folder, output_folder)
```
