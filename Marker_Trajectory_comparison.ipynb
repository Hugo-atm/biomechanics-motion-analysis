```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
file_paths = ['D:\Tugas Akhir\Subject TA\S17\Marker\S17MARKER_PIXEL_FILTERED.csv', 'D:\Tugas Akhir\Subject TA\S17\YOLOV7\S17_YOLOV7_FILTERED.csv', 'D:\Tugas Akhir\Subject TA\S17\OpenPose csv\S17_FR_OPENPOSE_INTERPOLATED_FILTERED.csv']
```

```python
file_paths
```

```python
fps = 120  
duration_seconds = 12  
final_frame = fps * duration_seconds
x_index = range(final_frame)
```

```python
# Create a figure for plottin, 12))
plt.figure(figsize=(20, 15))

# Define labels for legends based on your file names or descriptions
labels = ['Marker method', 'Yolov7', 'OpenPose']  # Replace with appropriate labels if needed

# Loop through each file and plot data
for idx, file_path in enumerate(file_paths):
    df = pd.read_csv(file_path)
    subset_df = df.iloc[:final_frame]

    # Marker 1
    plt.subplot(6, 1, 1)
    plt.plot(x_index, subset_df["Marker1x"], label=labels[idx])
    plt.title("Marker 1 X-Coordinate vs. Frame")
    plt.ylabel("X-Coordinate")

    plt.subplot(6, 1, 2)
    plt.plot(x_index, subset_df["Marker1y"], label=labels[idx])
    plt.title("Marker 1 Y-Coordinate vs. Frame")
    plt.ylabel("Y-Coordinate")

    # Marker 2
    plt.subplot(6, 1, 3)
    plt.plot(x_index, subset_df["Marker2x"], label=labels[idx])
    plt.title("Marker 2 X-Coordinate vs. Frame")
    plt.ylabel("X-Coordinate")

    plt.subplot(6, 1, 4)
    plt.plot(x_index, subset_df["Marker2y"], label=labels[idx])
    plt.title("Marker 2 Y-Coordinate vs. Frame")
    plt.ylabel("Y-Coordinate")

    # Marker 3
    plt.subplot(6, 1, 5)
    plt.plot(x_index, subset_df["Marker3x"], label=labels[idx])
    plt.title("Marker 3 X-Coordinate vs. Frame")
    plt.ylabel("X-Coordinate")

    plt.subplot(6, 1, 6)
    plt.plot(x_index, subset_df["Marker3y"], label=labels[idx])
    plt.title("Marker 3 Y-Coordinate vs. Frame")
    plt.ylabel("Y-Coordinate")
    plt.xlabel("Frame number")

# Adding legends to each subplot
for i in range(1, 7):
    plt.subplot(6, 1, i)
    plt.legend()
    plt.grid()

plt.tight_layout()
plt.show()
```
